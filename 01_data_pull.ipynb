```python
!pip install mp-api -q

from google.colab import userdata
from mp_api.client import MPRester

API_KEY = userdata.get('MP_API_KEY')

with MPRester(API_KEY) as mpr:
    docs = mpr.materials.summary.search(
        formula="**O3",
        fields=["material_id", "formula_pretty", "band_gap"],
        num_chunks=1, chunk_size=10
    )

for d in docs:
    print(d.formula_pretty, d.band_gap)
```

```python
import pandas as pd

df = pd.DataFrame([d.model_dump() for d in docs])
df.to_csv("perovskite_bandgaps.csv", index=False)
print(df.shape)
df.head()
```

```python
import pandas as pd

with MPRester(API_KEY) as mpr:
    docs = mpr.materials.summary.search(
        formula="**O3",
        fields=["material_id", "formula_pretty", "band_gap",
                "energy_above_hull", "is_stable"]
    )

df = pd.DataFrame([d.model_dump() for d in docs])
df = df[["material_id", "formula_pretty", "band_gap",
         "energy_above_hull", "is_stable"]]

df.to_csv("perovskite_bandgaps.csv", index=False)
print(df.shape)
print(df.head())
```
